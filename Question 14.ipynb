## Load Libraries 

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

## Load Obesity Dataset from Kaggle and Review 

```python
Obesity_CSV = "ObesityDataSet_raw_and_data_sinthetic.csv"
df = pd.read_csv(Obesity_CSV)
```

```python
df.head(10)
```

## Select Chosen Features

```python
Features = ['Height', 'Age', 'Gender', 'Weight', 'SMOKE']
Features_subset = df[Features].dropna()

Encoded_subset = Features_subset.copy()
ENC_gender = LabelEncoder()
ENC_smoke = LabelEncoder()

Encoded_subset['Gender'] = ENC_gender.fit_transform(Encoded_subset['Gender'])
Encoded_subset['SMOKE'] = ENC_smoke.fit_transform(Encoded_subset['SMOKE'])

Encoded_subset.head(10)
```

```python
x = Encoded_subset[['Height', 'Age', 'Gender', 'SMOKE']]
y = Encoded_subset['Weight']
```

## Split Data (30% Test with 70% Train)

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
```

## Train the Model

```python
lr_model = LinearRegression()
lr_model.fit(x_train, y_train)
```

## Test Set Prediction and Mean Square Error Generation

```python
y_pred = lr_model.predict(x_test)

Mean_Square = mean_squared_error(y_test, y_pred)
print(round(Mean_Square, 3))
```
